### Run script

```python
import sys
import time

try:
    drawer.main()

except KeyboardInterrupt:
    print("\n[!] Execution interrupted by user.")

finally:
    controller = None
    for attr_name in dir(drawer):
        attr = getattr(drawer, attr_name)
        if attr.__class__.__name__ == 'GRBLController':
            controller = attr
            break

    if controller and hasattr(controller, 'ser') and controller.ser:
        try:
            if controller.ser.is_open:
                controller.ser.reset_input_buffer()   # Discard broken GRBL strings
                controller.ser.reset_output_buffer()  # Wipe pending movement lines
                controller.ser.close()                # Unlock the physical USB port
                print("-> GRBLController serial port successfully freed.")
        except Exception as e:
            print(f"-> Error cleaning up serial port: {e}")

    time.sleep(0.5)
    print("Cleanup complete. Cell is ready to rerun.")
```
